```python
# Cell 2 — Import Required Libraries

from pathlib import Path

from astroquery.mast import Observations
from astropy.io import fits

from astropy.coordinates import SkyCoord
import astropy.units as u

print("Libraries imported successfully.")
```

```python
# Cell 3 — Create a download directory

data_dir = Path("../data/raw")

data_dir.mkdir(parents=True, exist_ok=True)

print(data_dir.resolve())
```

```python
# Cell 4 — Query MAST archive

obs = Observations.query_object(
    "Hubble Ultra Deep Field",
    radius="0.02 deg"
)

print(f"Observations found: {len(obs)}")

obs[:10]
```

```python
# Cell 5 - Display HST Query Results ==================

coords = SkyCoord(
    "03h32m39s -27d47m29s",
    frame="icrs"
)

obs = Observations.query_criteria(
    coordinates=coords,
    radius=0.02 * u.deg,
    obs_collection="HST"
)

print(f"HST observations found: {len(obs)}")

obs[:10]
```

```python
# Cell 6 - Inspect Available Metadata Columns

print(obs.colnames)
```

```python
# Cell 7 - Inspect Selected Observation Metadata ===========================

obs[
    "target_name",
    "instrument_name",
    "filters",
    "proposal_id",
    "t_exptime"
][:20]
```

Inspection of the returned metadata shows that the HST archive contains both scientific observations and instrument calibration data. Target name, proposal ID, filter, and exposure time provide useful context for distinguishing deep astronomical observations from calibration frames. The table above contains examples of both.

```python
# Cell 8 - Dataset Summary ==========================

print(f"Number of observations : {len(obs)}")

print("\nUnique instruments:")
print(sorted(set(obs["instrument_name"])))
```

```python
# Cell 9 — Save Search Results to data/catalogs

obs.write(
    "../data/catalogs/hudf_search.ecsv",
    format="ascii.ecsv",
    overwrite=True
)

print("Saved search results.")
```
