```python
# ==========================================
# Cell 2 — Imports
# ==========================================

import numpy as np
import matplotlib.pyplot as plt

from astropy.table import Table
from astropy.coordinates import SkyCoord
import astropy.units as u

from pyvo.dal import TAPService
```

```python
# ==========================================
# Cell 3 — Load the Target Galaxy
# ==========================================

from astropy.table import Table
from astropy.coordinates import SkyCoord
import astropy.units as u

# Path to the Einstein master catalog
catalog_file = "../data/catalogs/einstein_master_catalog.ecsv"

# Load the catalog
catalog = Table.read(catalog_file, format="ascii.ecsv")

# Select the target galaxy
target_id = 39
target = catalog[catalog["label"] == target_id][0]

# Coordinates from the Einstein catalog
target_ra = float(target["ra_deg"])
target_dec = float(target["dec_deg"])

# Properties established in Notebook 20
target_name = "EIS J033238.61-274631.6"
target_type = "Galaxy"
target_z = 0.621743

# Sky coordinate object for later matching
target_coord = SkyCoord(
    ra=target_ra * u.deg,
    dec=target_dec * u.deg,
    frame="icrs"
)

print("=" * 50)
print("Target Galaxy")
print("=" * 50)
print(f"Einstein Label : {target_id}")
print(f"Catalog Name   : {target_name}")
print(f"Object Type    : {target_type}")
print(f"Redshift       : {target_z:.6f}")
print(f"RA             : {target_ra:.6f} deg")
print(f"Dec            : {target_dec:.6f} deg")
```

```python
# ---------------------------------------------------------
# Cell 4 - Connect to CDS TAP
# ---------------------------------------------------------

tap = TAPService(
    "https://tapvizier.cds.unistra.fr/TAPVizieR/tap"
)

print("Connected to CDS TAP.")
```

```python
# =========================================================
# Cell 5 — Define the ALMA Survey
# =========================================================

# This notebook cross-matches Object 39 against a published
# ALMA survey of the GOODS-South field.

survey_name = "GOODS-South ALMA Survey"

# Search radius for positional matching
search_radius = 5.0 * u.arcsec

print("=" * 50)
print("ALMA Survey")
print("=" * 50)
print(f"Survey        : {survey_name}")
print(f"Search Radius : {search_radius}")
```

```python
# =========================================================
# Cell 6 — Query the ALMA Science Archive
# =========================================================

from pyvo.dal import TAPService

# Connect to the ALMA TAP service
alma_service = TAPService("https://almascience.eso.org/tap")

# Search radius in degrees
radius_deg = search_radius.to(u.deg).value

query = f"""
SELECT
    target_name,
    s_ra,
    s_dec,
    obs_id,
    proposal_id,
    obs_collection,
    facility_name,
    instrument_name,
    em_min,
    em_max,
    t_exptime,
    access_url
FROM ivoa.obscore
WHERE
    INTERSECTS(
        CIRCLE('ICRS',
               {target_ra},
               {target_dec},
               {radius_deg}),
        s_region
    ) = 1
"""

print("=" * 60)
print("Searching the ALMA Science Archive...")
print("=" * 60)

results = alma_service.search(query)

alma_table = results.to_table()

print(f"ALMA archive rows returned: {len(alma_table)}")

alma_table
```

```python
# =========================================================
# Cell 7 — Compute Angular Separations
# =========================================================

from astropy.coordinates import SkyCoord

# Build SkyCoord objects directly from the table columns.
# The ALMA TAP service already provides the coordinate units.
alma_coords = SkyCoord(
    ra=alma_table["s_ra"],
    dec=alma_table["s_dec"],
    frame="icrs"
)

# Compute angular separation (arcseconds)
alma_table["separation_arcsec"] = (
    target_coord.separation(alma_coords).arcsecond
)

# Sort by nearest observation
alma_table.sort("separation_arcsec")

print("=" * 60)
print("Nearest ALMA observations")
print("=" * 60)

display_columns = [
    "target_name",
    "proposal_id",
    "obs_id",
    "separation_arcsec",
    "t_exptime",
    "em_min",
    "em_max",
]

alma_table[display_columns][:10]
```

```python
# =========================================================
# Cell 8 — Summarize Unique ALMA Observations
# =========================================================

alma_summary = (
    alma_table[
        "target_name",
        "proposal_id",
        "separation_arcsec",
        "t_exptime"
    ]
    .group_by(["proposal_id", "target_name"])
)

alma_summary = alma_summary.groups.aggregate(np.min)

alma_summary.sort("separation_arcsec")

alma_summary
```

```python
# =========================================================
# Cell 9 — Rank ALMA Observing Programs
# =========================================================

from astropy.table import Table
import numpy as np

# Make a copy so the original summary remains unchanged
ranked_programs = alma_summary.copy()

# Convert exposure time from seconds to hours
ranked_programs["Exposure_hr"] = ranked_programs["t_exptime"] / 3600.0

# Round values for readability
ranked_programs["Exposure_hr"] = np.round(ranked_programs["Exposure_hr"], 2)
ranked_programs["separation_arcsec"] = np.round(
    ranked_programs["separation_arcsec"], 3
)

ranked_programs["Year"] = [
    int(pid.split(".")[0])
    for pid in ranked_programs["proposal_id"]
]

# Add ranking column
ranked_programs["Rank"] = np.arange(1, len(ranked_programs) + 1)

# Reorder columns
ranked_programs = ranked_programs[
    "Rank",
    "Year",
    "target_name",
    "proposal_id",
    "separation_arcsec",
    "Exposure_hr",
]

print("=" * 70)
print("Nearest ALMA Observing Programs")
print("=" * 70)
print()

ranked_programs
```

```python
# =========================================================
# Cell 10 — Einstein Assessment
# =========================================================

n_programs = len(ranked_programs)

closest = ranked_programs[0]

print("=" * 70)
print("Einstein Assessment")
print("=" * 70)
print()

print(f"Object: {target_name}")
print(f"Type: {target_type}")
print(f"Redshift: {target_z:.6f}")
print()

print(
    f"The ALMA Science Archive contains {n_programs} unique "
    f"observing programs within the search region."
)
print()

print(
    f"The nearest observation is '{closest['target_name']}' "
    f"(Proposal {closest['proposal_id']}), "
    f"located {closest['separation_arcsec']:.3f} arcseconds "
    f"from the target with an integration time of "
    f"{closest['Exposure_hr']:.2f} hours."
)
print()

print(
    "Object 39 lies within the GOODS-South / Hubble Ultra Deep Field "
    "region, one of the most extensively observed extragalactic fields "
    "in astronomy. Multiple ALMA observing programs from 2012–2025 "
    "cover this region, making the object an excellent candidate for "
    "future archival millimetre and submillimetre analysis."
)
```
