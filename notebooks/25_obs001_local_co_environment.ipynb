```python
# ==========================================================
# Phase 1A
# Load the ASPECS CO Catalog
# ==========================================================

co_table = Table.read(CATALOGS / "aspecs_co_catalog.ecsv")

print(co_table)
```

```python
# ==========================================================
# Phase 1B
# Create Sky Coordinates
# ==========================================================

co_coords = SkyCoord(
    co_table["ra"],
    co_table["dec"],
    unit=(u.hourangle, u.deg),
    frame="icrs",
)

print(co_coords[:5])
```

```python
# ==========================================================
# Phase 1B
# Compute Angular Separations from C19
# ==========================================================

# Compute separations
co_table["sep_arcsec"] = c19.separation(co_coords).arcsecond

# Sort by increasing separation
co_table.sort("sep_arcsec")

# Display the most relevant columns
co_table["id", "ra", "dec", "z", "sep_arcsec"]
```

```python
# ==========================================================
# Phase 1B
# Nearby CO Sources
# ==========================================================

within10 = co_table[co_table["sep_arcsec"] <= 10]
within20 = co_table[co_table["sep_arcsec"] <= 20]
within30 = co_table[co_table["sep_arcsec"] <= 30]

print(f"Within 10 arcsec : {len(within10)}")
print(f"Within 20 arcsec : {len(within20)}")
print(f"Within 30 arcsec : {len(within30)}")
```

```python
print("\nWithin 10 arcsec")
print(within10["id", "z", "sep_arcsec"])

print("\nWithin 20 arcsec")
print(within20["id", "z", "sep_arcsec"])

print("\nWithin 30 arcsec")
print(within30["id", "z", "sep_arcsec"])
```

```python
# ==========================================================
# Phase 1C
# Redshift Differences
# ==========================================================

co_table["delta_z"] = abs(co_table["z"] - c19_redshift)

co_table["likely_companion"] = co_table["delta_z"] < 0.05

co_table["id", "z", "delta_z", "sep_arcsec", "likely_companion"]
```

## Phase 1C Summary

The ASPECS CO catalog was searched for neighboring CO detections around
C19.

Within 30 arcseconds, no additional CO source has a redshift consistent
with C19 (Δz < 0.05).

The nearest redshift-consistent source is CO ID 1, located 31.9
arcseconds from C19 with Δz = 0.031.

These results suggest that C19 is not part of a compact CO-selected
pair within the ASPECS sample, although a wider-scale association
cannot yet be excluded.

```python
# ==========================================================
# Phase 1C
# Classify CO Neighbors
# ==========================================================

def classify_neighbor(row):
    if row["sep_arcsec"] <= 10 and row["delta_z"] < 0.05:
        return "Target (C19)"
    elif row["sep_arcsec"] <= 30 and row["delta_z"] < 0.05:
        return "Local Companion"
    elif row["delta_z"] < 0.05:
        return "Candidate Same-Redshift Source"
    else:
        return "Different Redshift"

co_table["association_status"] = [
    classify_neighbor(row) for row in co_table
]

co_table["id", "z", "delta_z", "sep_arcsec", "association_status"]
```

```python
# ==========================================================
# Phase 1C
# Save CO Neighbor Analysis
# ==========================================================

co_table.write(
    CATALOGS / "obs001_co_neighbor_analysis.ecsv",
    overwrite=True
)

print("Saved:", CATALOGS / "obs001_co_neighbor_analysis.ecsv")
```

## Phase 1C Results

The ASPECS CO catalog was searched for neighboring CO detections
around ASPECS C19.

### Measurements

- Four CO detections lie within 10 arcseconds.
- Four CO detections lie within 20 arcseconds.
- Five CO detections lie within 30 arcseconds.

Only the target galaxy (CO ID 12) has Δz = 0.

CO ID 1 has Δz = 0.031 but lies 31.9 arcseconds from C19.

All remaining CO detections have substantially different redshifts.

### Preliminary Interpretation

No additional CO-selected galaxy within 30 arcseconds satisfies the
current same-redshift criterion (Δz < 0.05).

CO ID 1 lies just outside the adopted search radius and is retained
as a candidate for investigation during the projected-separation
analysis.

No conclusions regarding physical association are drawn at this stage.

```python
# ==========================================================
# Phase 1C
# Create Analysis Table
# ==========================================================

analysis_table = co_table.copy()

analysis_table.write(
    CATALOGS / "obs001_co_neighbor_analysis.ecsv",
    overwrite=True
)

print("Saved:", CATALOGS / "obs001_co_neighbor_analysis.ecsv")
```

```python
# ==========================================================
# Phase 1D
# Physical Scale at the Redshift of C19
# ==========================================================

scale = Planck18.kpc_proper_per_arcmin(c19_redshift)

print(scale)
```

```python
kpc_per_arcsec = scale / 60

print(kpc_per_arcsec)
```

```python
# ==========================================================
# Phase 1D
# Projected Physical Separations
# ==========================================================

analysis_table["projected_kpc"] = (
    analysis_table["sep_arcsec"] *
    kpc_per_arcsec.value
)

analysis_table["id",
               "z",
               "sep_arcsec",
               "projected_kpc",
               "association_status"]
```

```python
# ==========================================================
# Phase 1D
# Velocity Differences
# ==========================================================

from astropy.constants import c

analysis_table["delta_v_kms"] = (
    c.to("km/s").value *
    analysis_table["delta_z"] /
    (1 + c19_redshift)
)

analysis_table["id",
               "z",
               "delta_z",
               "delta_v_kms",
               "projected_kpc"]
```
