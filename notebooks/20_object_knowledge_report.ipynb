# Section 2 — Multiwavelength Evidence
| Wavelength    | Evidence   |
| ------------- | ---------- |
| Optical       | EIS galaxy |
| Infrared      | WISE       |
| Submillimeter | ASPECS     |
| ALMA          | ALMA-HUDF  |
| GOODS         | GOODS-CDFS |


```python
# ============================================================
# Cell 6 - Classify Catalog Surveys
# ============================================================

def classify_catalog(name):
    """
    Classify catalog entries by their primary observing wavelength
    or survey type.
    """

    name = str(name)

    # ------------------------
    # Optical / Hubble
    # ------------------------
    if (
        name.startswith("EIS")
        or name.startswith("PEARS")
        or name.startswith("CANDELS")
        or name.startswith("ACS-GC")
        or name.startswith("UDF")
        or name.startswith("UDF10")
        or name.startswith("UDF12")
        or name.startswith("HRG14")
        or "CVU" in name
    ):
        return "Optical"

    # ------------------------
    # Infrared
    # ------------------------
    elif (
        name.startswith("WISE")
        or name.startswith("SSTSL2")
        or name.startswith("GOODS")
    ):
        return "Infrared"

    # ------------------------
    # Ultraviolet
    # ------------------------
    elif name.startswith("GALEX"):
        return "Ultraviolet"

    # ------------------------
    # X-ray
    # ------------------------
    elif (
        name.startswith("2CXO")
        or name.startswith("[XWB")
    ):
        return "X-ray"

    # ------------------------
    # Radio / Submillimeter
    # ------------------------
    elif (
        name.startswith("ASPECS")
        or name.startswith("SMM")
    ):
        return "Submillimeter"

    # ------------------------
    # Millimeter (ALMA)
    # ------------------------
    elif "ALMA" in name:
        return "Millimeter"

    # ------------------------
    # Everything else
    # ------------------------
    else:
        return "Other"


survey_type = [
    classify_catalog(name)
    for name in ned_table["Object Name"]
]

ned_table["Survey Type"] = survey_type

print("Survey classifications added.")
```

```python
# ============================================================
# Cell 7 - Multiwavelength Summary Table
# ============================================================

columns = [
    "Object Name",
    "Survey Type",
    "Type",
    "Redshift",
    "Einstein Separation (arcsec)"
]

survey_table = ned_table[columns].copy()

survey_table.sort("Einstein Separation (arcsec)")

survey_table
```

```python
# ============================================================
# Cell 8 - Count Observations by Survey Type
# ============================================================

from collections import Counter

counts = Counter(survey_table["Survey Type"])

print("=" * 70)
print("Multiwavelength Coverage")
print("=" * 70)

total = len(survey_table)

for survey, number in sorted(counts.items()):
    percent = 100 * number / total
    print(f"{survey:<25} {number:>3} ({percent:5.1f}%)")

print()
print(f"Total catalog entries: {total}")
```

```python
# ============================================================
# Cell 8A - Inspect Unclassified Catalogs
# ============================================================

unknown = survey_table[survey_table["Survey Type"] == "Other"]

print(f"Number of unclassified entries: {len(unknown)}\n")

for name in sorted(set(unknown["Object Name"])):
    print(name)
```

```python
# ============================================================
# Cell 9 - Einstein Interpretation
# ============================================================

print("=" * 70)
print("Einstein Interpretation")
print("=" * 70)

print()

print(
    "The target position is represented in multiple independent "
    "astronomical surveys spanning several wavelength regimes."
)

print()

for survey, number in sorted(counts.items()):
    percent = 100 * number / len(survey_table)
    print(f"• {survey:<18} {number:2d} catalog entries ({percent:4.1f}%)")

print()

print(
    "The predominance of optical detections is expected because "
    "the target lies within the Hubble Ultra Deep Field, one of "
    "the most extensively observed regions of the sky."
)

print()

print(
    "Infrared, ultraviolet, X-ray, and submillimeter detections "
    "provide complementary evidence that this object has been "
    "studied by numerous independent observing programs."
)

print()

print(f"""
Overall Assessment
------------------

Einstein Object {TARGET_LABEL} is confidently associated with the galaxy
{closest['Object Name']}.

The agreement between Einstein's measured position and multiple
independent catalog identifications across optical, infrared,
submillimeter, millimeter, ultraviolet, and X-ray surveys
provides strong evidence that this is a well-characterized
extragalactic object.
""")
```

```python
# ============================================================
# Cell 10 - Cosmology
# ============================================================

from astropy.cosmology import Planck18
import astropy.units as u

print("Using Planck18 cosmology.")
```

```python
# ============================================================
# Cell 11 - Calculate Physical Properties
# ============================================================

z = float(closest["Redshift"])

lookback_time = Planck18.lookback_time(z)
age_universe = Planck18.age(z)

luminosity_distance = Planck18.luminosity_distance(z)
comoving_distance = Planck18.comoving_distance(z)
angular_diameter_distance = Planck18.angular_diameter_distance(z)

print("Cosmological calculations complete.")
```

```python
# ============================================================
# Cell 12 - Physical Properties
# ============================================================

print("=" * 70)
print("Physical Properties")
print("=" * 70)

print(f"\nRedshift")
print("-" * 70)
print(f"z = {z:.6f}")

print("\nDistances")
print("-" * 70)

print(f"Comoving Distance        : {comoving_distance:.2f}")
print(f"Luminosity Distance      : {luminosity_distance:.2f}")
print(f"Angular Diameter Distance: {angular_diameter_distance:.2f}")

print("\nCosmic Time")
print("-" * 70)

print(f"Lookback Time            : {lookback_time:.2f}")
print(f"Age of Universe          : {age_universe:.2f}")
```

```python
# ============================================================
# Cell 13 - Einstein Interpretation
# ============================================================

print("=" * 70)
print("Cosmological Interpretation")
print("=" * 70)

print(f"""
The primary galaxy has a measured redshift of z = {z:.3f}.

According to the Planck18 cosmological model,
the light observed by the Hubble Space Telescope
has been travelling for approximately
{lookback_time.value:.2f} billion years.

When this light left the galaxy, the Universe
was approximately {age_universe.value:.2f}
billion years old.

This means Einstein is observing the galaxy
as it existed billions of years in the past,
providing a direct view of cosmic history.

The galaxy's comoving distance is
approximately {comoving_distance.value:.2f}
Gpc, while its luminosity distance is
approximately {luminosity_distance.value:.2f}
Gpc.
""")
```

```python
# ============================================================
# Cell 15 - Gather Confidence Metrics
# ============================================================

closest_sep = float(closest["Einstein Separation (arcsec)"])

num_catalogs = len(ned_table)

num_wavelengths = len(counts)

redshift_available = closest["Redshift"] is not np.ma.masked

print("Confidence metrics collected.")
```

```python
# ============================================================
# Cell 16 - Confidence Assessment
# ============================================================

score = 0

# Positional agreement
if closest_sep < 0.2:
    score += 4
elif closest_sep < 0.5:
    score += 3
elif closest_sep < 1.0:
    score += 2
else:
    score += 1

# Object classification available
if str(closest["Type"]) != "--":
    score += 2

# Redshift available
if redshift_available:
    score += 2

# Multiple independent wavelengths
if num_wavelengths >= 5:
    score += 2

# Large number of catalog detections
if num_catalogs >= 20:
    score += 2

print(f"Confidence Score = {score}")
```

```python
# ============================================================
# Cell 17 - Confidence Rating
# ============================================================

if score >= 11:
    confidence = "Very High"

elif score >= 9:
    confidence = "High"

elif score >= 6:
    confidence = "Moderate"

else:
    confidence = "Low"

print(f"Confidence Rating: {confidence}")
```

```python
# ============================================================
# Cell 18 - Einstein Confidence Report
# ============================================================

print("=" * 70)
print("Confidence Assessment")
print("=" * 70)

print()

print(f"Overall Confidence : {confidence}")
print(f"Confidence Score   : {score}/12")

print("\nEvidence")

print("-" * 70)

print(f"✓ Closest positional match : {closest_sep:.3f} arcsec")

print(f"✓ Object classified as     : {closest['Type']}")

if redshift_available:
    print(f"✓ Redshift available       : {z:.6f}")
else:
    print("• No redshift available")

print(f"✓ Catalog entries          : {num_catalogs}")

print(f"✓ Survey types             : {num_wavelengths}")

print()

print(
    "Based on positional agreement and the available catalog "
    "information, the primary counterpart is the strongest "
    "candidate identified in this NED search."
)

print(
    "No competing catalog object provides stronger evidence "
    "for association with the Einstein detection."
)
```
