```python
# Cell 2 — Import Required Libraries ===========

from pathlib import Path

from astropy.table import Table
from astroquery.mast import Observations
```

```python
# Cell 3 — Set Project Directories ===============

data_dir = Path("../data/raw")

print(data_dir.resolve())
```

```python
# Cell 4 - Load Saved Search Results ===============

obs = Table.read(
    "../data/catalogs/hudf_search.ecsv",
    format="ascii.ecsv"
)

print(f"Loaded {len(obs)} observations.")
```

```python
# Cell 5 - Select HUDF-DEEP-WFC3 ====================

hudf = obs[obs["target_name"] == "HUDF-DEEP-WFC3"]

print(f"Found {len(hudf)} observations")

hudf[:10]
```

```python
# Cell 6 - Retrieve Available Data Products =========

products = Observations.get_product_list(hudf)

print(f"Products found: {len(products)}")
```
