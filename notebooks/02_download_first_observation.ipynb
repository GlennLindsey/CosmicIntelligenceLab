### Research Question

How can we retrieve, verify, and store a specific Hubble observation for later AI analysis?

### Objectives

- Select a Hubble observation
- Retrieve its associated data products
- Download one FITS image
- Store it in `data/raw`
- Verify the download

```python
# Cell 1A =============================

from astropy.table import Table

obs = Table.read("../data/catalogs/hudf_search.ecsv",
                 format="ascii.ecsv")
```

```python
# Cell 2 — Import Required Libraries ===========

from pathlib import Path

from astropy.table import Table
from astroquery.mast import Observations
```

```python
# Cell 3 — Set Project Directories ===============

data_dir = Path("../data/raw")

print(data_dir.resolve())
```

```python
# Cell 4 - Load Saved Search Results ===============

obs = Table.read(
    "../data/catalogs/hudf_search.ecsv",
    format="ascii.ecsv"
)

print(f"Loaded {len(obs)} observations.")
```

```python
# Cell 5 - Select HUDF-DEEP-WFC3 ====================

hudf = obs[obs["target_name"] == "HUDF-DEEP-WFC3"]

print(f"Found {len(hudf)} observations")

hudf[:10]
```

```python
# Cell 5a - Inspect HUDF Observations =========================

hudf["obsid", "obs_id", "instrument_name", "filters"][:20]
```

```python
# Cell 6 - Select First Science Observation ==================

one_obs = hudf[7:8]

print(one_obs)
print(f"\nSelected observation: {len(one_obs)}")
```

```python
# Cell 7 - Retrieve Products ========================

products = Observations.get_product_list(one_obs)

print(f"Products found: {len(products)}")
```

```python
# Cell 8 - Inspect Available Products ==========================

products[
    "productFilename",
    "productSubGroupDescription",
    "productType",
    "size"
][:25]
```

```python
# Cell 9 - Select the FLT Product =============================

flt = products[
    products["productSubGroupDescription"] == "FLT"
]

print(f"FLT products found: {len(flt)}")

flt["productFilename", "size"]
```

```python
manifest = Observations.download_products(
    flt[:1],
    download_dir="../data/raw",
    mrp_only=False
)

manifest
```

```python
# Cell 11 - Verify Download =========================

from pathlib import Path

fits_file = Path("../data/raw/mastDownload/HST/ib5x1fetq/ib5x1fetq_flt.fits")

print(f"Exists : {fits_file.exists()}")

if fits_file.exists():
    print(f"Size   : {fits_file.stat().st_size:,} bytes")
    print(f"File   : {fits_file.name}")
```
