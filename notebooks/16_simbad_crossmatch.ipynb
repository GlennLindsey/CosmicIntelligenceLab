````python
# ============================================================
# Cell 2 - Import Libraries
# ============================================================

from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt

from astropy.table import Table
from astropy.coordinates import SkyCoord

import astropy.units as u

from astroquery.simbad import Simbad
````

````python
# ============================================================
# Cell 3 - Load Einstein Master Catalog
# ============================================================

catalog_file = Path("../data/catalogs/einstein_master_catalog.ecsv")

catalog = Table.read(catalog_file)

print("=" * 60)
print("Einstein Master Catalog")
print("=" * 60)

print(f"Objects loaded : {len(catalog)}")

catalog[:5]
````

````python
# ============================================================
# Cell 4 - Configure SIMBAD
# ============================================================

custom_simbad = Simbad()

custom_simbad.add_votable_fields(
    "otype",
    "ra",
    "dec"
)

print("SIMBAD connection configured.")
````

````python
# ============================================================
# Cell 5 - Select Einstein's First Target
# ============================================================

target = catalog[catalog["label"] == 45][0]

print("=" * 60)
print("Einstein's First Target")
print("=" * 60)

print(f"Label : {target['label']}")
print(f"Flux  : {target['segment_flux']:.2f}")
print(f"RA    : {target['ra_deg']:.6f}°")
print(f"Dec   : {target['dec_deg']:.6f}°")
````
