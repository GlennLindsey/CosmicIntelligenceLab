````python
# Cell 1a - Investigation target

TARGET_LABEL = 45
````

````python
# ============================================================
# Cell 3 - Load Einstein Master Catalog
# ============================================================

from pathlib import Path
from astropy.table import Table

catalog_file = Path("../data/catalogs/einstein_master_catalog.ecsv")

catalog = Table.read(catalog_file)

print("=" * 60)
print("Einstein Master Catalog")
print("=" * 60)

print(f"Objects loaded : {len(catalog)}")
````

````python
# ============================================================
# Cell 4 - Select Einstein's First Target
# ============================================================

target = catalog[catalog["label"] == TARGET_LABEL][0]

print("=" * 60)
print("Einstein's First Target")
print("=" * 60)

print(f"Label : {target['label']}")
print(f"Flux  : {target['segment_flux']:.2f}")
print(f"RA    : {target['ra_deg']:.6f}°")
print(f"Dec   : {target['dec_deg']:.6f}°")
````

````python
# ============================================================
# Cell 5 - Configure SIMBAD
# ============================================================

from astroquery.simbad import Simbad

custom_simbad = Simbad()

custom_simbad.add_votable_fields(
    "otype",
    "ra",
    "dec"
)

print("SIMBAD configured.")
````

````python
# ============================================================
# Cell 6 - Einstein's First SIMBAD Query
# ============================================================

from astropy.coordinates import SkyCoord
import astropy.units as u

# Build SkyCoord object
coord = SkyCoord(
    ra=target["ra_deg"] * u.deg,
    dec=target["dec_deg"] * u.deg,
    frame="icrs"
)

# Search within 5 arcseconds
result = custom_simbad.query_region(
    coord,
    radius=5 * u.arcsec
)

print("=" * 60)
print("SIMBAD Query Result")
print("=" * 60)

if result is None:
    print("No object found.")
else:
    print(result)
````

````python
# ============================================================
# Inspect the SIMBAD result table
# ============================================================

print(result)

print("\nColumn names:")
print(result.colnames)
````

````python
# ============================================================
# Cell 7 - Einstein Interprets the Result
# ============================================================

if result is None:
    print("No SIMBAD object found.")

else:

    obj = result[0]

    print("=" * 60)
    print("Einstein's First Identification")
    print("=" * 60)

    print(f"Object Name : {obj['main_id']}")
    print(f"Object Type : {obj['otype']}")
    print(f"RA          : {obj['ra']}")
    print(f"Dec         : {obj['dec']}")
````

## First Scientific Identification

Einstein has successfully identified its first AI-discovered object
using the SIMBAD astronomical database.

Object **Label 45**, originally detected in the Hubble image and
classified through machine learning, has been matched with the
published galaxy **[RTG2015] 24643**.

This represents the first successful integration of Einstein's own
measurements with humanity's accumulated astronomical knowledge.

From this point forward, Einstein is no longer limited to measuring
objects—it can begin learning what astronomers already know about them.

````python
# ============================================================
# Cell 8 - Einstein's First Scientific Object Profile
# ============================================================

profile = {
    "Einstein Label": int(target["label"]),
    "SIMBAD Name": str(obj["main_id"]),
    "Object Type": str(obj["otype"]),

    # Einstein measurements
    "Measured Flux": float(target["segment_flux"]),
    "Area (pixels)": int(target["area"]),
    "Semimajor Axis (pixels)": float(target["semimajor_axis"]),
    "Semiminor Axis (pixels)": float(target["semiminor_axis"]),
    "Orientation (deg)": float(target["orientation"]),
    "Eccentricity": float(target["eccentricity"]),

    # Coordinates
    "Einstein RA (deg)": float(target["ra_deg"]),
    "Einstein Dec (deg)": float(target["dec_deg"]),
    "SIMBAD RA": str(obj["ra"]),
    "SIMBAD Dec": str(obj["dec"]),
}

print("=" * 65)
print("Einstein Scientific Object Profile")
print("=" * 65)

for key, value in profile.items():
    print(f"{key:<28} : {value}")
````

````python
# ============================================================
# Cell 9 - Einstein's Interpretation
# ============================================================

print("=" * 60)
print("Einstein's Scientific Interpretation")
print("=" * 60)

print(f"""
Object {profile['Einstein Label']} has been successfully matched
with the SIMBAD database.

Published Identifier:
    {profile['SIMBAD Name']}

Classification:
    {profile['Object Type']} (Galaxy)

Einstein's measurements indicate an extended object with:

    Flux              : {profile['Measured Flux']:.2f}
    Area              : {profile['Area (pixels)']} pixels
    Semimajor Axis    : {profile['Semimajor Axis (pixels)']:.2f}
    Semiminor Axis    : {profile['Semiminor Axis (pixels)']:.2f}
    Orientation       : {profile['Orientation (deg)']:.2f}°
    Eccentricity      : {profile['Eccentricity']:.3f}

Conclusion

Einstein has successfully combined image-derived measurements with information 
retrieved from the SIMBAD astronomical database to produce its first Scientific 
Object Profile.

The identification of Object 45 as the galaxy [RTG2015] 24643 demonstrates that 
AI-generated detections can be linked directly to the astronomical literature, 
providing a foundation for richer scientific investigations in future notebooks.

This marks Einstein's first successful Scientific Object Profile.
""")
````

````python
# ============================================================
# Cell 10 - Save Scientific Object Profile
# ============================================================

from astropy.table import Table
from pathlib import Path

profile_table = Table(rows=[profile])

output_dir = Path("../data/catalogs")
output_dir.mkdir(parents=True, exist_ok=True)

outfile = output_dir / "object45_profile.ecsv"

profile_table.write(outfile, format="ascii.ecsv", overwrite=True)

print(f"Scientific profile saved to:\n{outfile}")
````
